**Finding Outliers**

```python
plt.figure(figsize= (20, 16))
plt.suptitle('Graphical Analysis Of Continuous Variable')

plt.subplot(3, 2, 1)
sns.boxplot(data= data, x='Age')
plt.title('Age Distribution')

plt.subplot(3, 2, 2)
sns.boxplot(data= data, x='Education')
plt.title('Education Distribution')

plt.subplot(3, 2, 3)
sns.boxplot(data= data, x='Usage')
plt.title('Usage Distribution')

plt.subplot(3, 2, 4)
sns.boxplot(data= data, x='Fitness')
plt.title('Fitness Distribution')

plt.subplot(3, 2, 5)
sns.boxplot(data= data, x='Income')
plt.title('Income Distribution')

plt.subplot(3, 2, 6)
sns.boxplot(data= data, x='Miles')
plt.title('Miles Distribution')

plt.show()
```

**Treatement Of Outliers**

**Clipping The Data Between 5 Percentile and 95 Percentile**

**NOTE:**
*Certainly, we acknowledge the presence of outliers in our dataset. However, removing them entirely can result in a loss of valuable information, especially given the relatively small size of our dataset, which consists of only 180 rows. Therefore, instead of outright removal, we'll opt for a clipped approach, wherein the data will be adjusted to fall within a certain range (between the 5th and 95th percentiles).*


*Please note that although the code provided demonstrates outlier treatment by clipping the data, this adjustment has not yet been incorporated into the subsequent analysis.*

```python
data_copy= data.copy()

Outliers_col=['Age','Education','Usage','Fitness', 'Income', 'Miles']
for col in Outliers_col:
    p5 = np.percentile(data_copy[col], 5)
    p95 = np.percentile(data_copy[col], 95)
    data_copy[col] = np.clip(data_copy[col], p5, p95)
```

**After Treatement Of Outliers**

**After Clipping The Data Between 5 percentile and 95 percentile**

```python
plt.figure(figsize= (20, 16))
plt.suptitle('After Clipping The Data Between 5 percentile and 95 percentile')

plt.subplot(3, 2, 1)
sns.boxplot(data= data_copy, x='Age')
plt.title('Age Distribution')

plt.subplot(3, 2, 2)
sns.boxplot(data= data_copy, x='Education')
plt.title('Education Distribution')

plt.subplot(3, 2, 3)
sns.boxplot(data= data_copy, x='Usage')
plt.title('Usage Distribution')

plt.subplot(3, 2, 4)
sns.boxplot(data= data_copy, x='Fitness')
plt.title('Fitness Distribution')

plt.subplot(3, 2, 5)
sns.boxplot(data= data_copy, x='Income')
plt.title('Income Distribution')

plt.subplot(3, 2, 6)
sns.boxplot(data= data_copy, x='Miles')
plt.title('Miles Distribution')

plt.show()
```

*>**Insight:**We've identified outliers within our dataset and have taken steps to address them. The data has been adjusted, ensuring that values now fall within a range defined by the 5th and 95th percentiles.*

**Converting Ages, Education, Fitness, Usage, Incomes and Miles To Category/Bins For Better Analysis.**

```python
age_rng = [0,21,35,45,float("inf")]
bin_lbl = ["Teen","Adult","Middle Aged","Elder"]
data['Age_cat'] = pd.cut(data['Age'],bins = age_rng, labels = bin_lbl)

edu_bins = [0,13,18,float('inf')]
edu_lbl = ['Low Edu', 'Moderate Edu', 'High Edu']
data["Education_cat"] = pd.cut(data["Education"], bins = edu_bins, labels = edu_lbl)

bin_mils = [0,50,100,200,float('inf')]
mils_lbl = ["Light Activity", "Moderate Activity", "Active Lifestyle", "Fitness Enthusiast"]
data["Miles_cat"] = pd.cut(data["Miles"], bins = bin_mils, labels = mils_lbl)

bin_inc = [0,40000,60000,80000,float("inf")]
inc_lbl = ["Low", "Moderate", "High Income", "Very High"]
data["Income_cat"] = pd.cut(data["Income"], bins = bin_inc, labels = inc_lbl)

data['Fitness_cat']= data.Fitness
data["Fitness_cat"].replace({1:"Poor Shape", 2:"Bad Shape", 3:"Average Shape", 4:"Good Shape", 5:"Excellent Shape"}, inplace=True)

use_bins = [0,3,5,float('inf')]
use_lbl = ['Low Usage', 'Moderate Usage', 'High Usage']
data["Usage_cat"] = pd.cut(data["Usage"], bins = use_bins, labels = use_lbl)
```

*>**Insight:**Here are the bins defined for converting ages, education levels, fitness ratings, incomes, and miles into categories for better analysis:*

Age Categories:

Teen: Ages 0-21
